```python
import requests
import pandas as pd
import time
```

```python
# TMDB API Key
import os
API_KEY = os.environ.get("b8cd668a03ee6296f94919995b52d187")

# Base URLs
NOW_PLAYING_URL = "https://api.themoviedb.org/3/movie/now_playing"
DETAIL_URL = "https://api.themoviedb.org/3/movie/{movie_id}"
CREDITS_URL = "https://api.themoviedb.org/3/movie/{movie_id}/credits"
KEYWORDS_URL = "https://api.themoviedb.org/3/movie/{movie_id}/keywords"

# Param_dict
params = {
    "api_key": API_KEY,
    "language": "en-US",
    "page": 1
}

all_movies = []

# Fetch first page
response = requests.get(NOW_PLAYING_URL, params = params)

if response.status_code == 200:
  data = response.json()
  total_pages = data.get("total_pages", 1)
  all_movies.extend(data.get("results", []))

else:
  raise Exception("Failed to fetch first page of now playing movies.")

# Fetch remaining pages
for page in range(2, total_pages + 1):
  params["page"] = page
  response = requests.get(NOW_PLAYING_URL, params = params)
  if response.status_code == 200:
    data = response.json()
    all_movies.extend(data.get("results", []))
  else:
    print(f"Failed to fetch page {page}.")
  time.sleep(0.2)

print(f"✅ Retrieved {len(all_movies)} now playing movies.")

# Step 2: Fetch full details, credits, and keywords for each movie
movie_data = []

for movie in all_movies:
  movie_id = movie.get("id")
  if not movie_id:
    continue

  # Detail, credits, and keywords requests
  detail_resp = requests.get(DETAIL_URL.format(movie_id = movie_id), params = {"api_key": API_KEY, "language": "en-US"})
  credits_resp = requests.get(CREDITS_URL.format(movie_id = movie_id), params = {"api_key": API_KEY, "language": "en-US"})
  keywords_resp = requests.get(KEYWORDS_URL.format(movie_id = movie_id), params = {"api_key": API_KEY, "language": "en-US"})

  if detail_resp.status_code == 200 and credits_resp.status_code == 200 and keywords_resp.status_code == 200:
    detail = detail_resp.json()
    credits = credits_resp.json()
    keywords = keywords_resp.json()
    movie_data.append({
        "id": movie_id,
        "title": detail.get("title"),
        "details_raw": detail,
        "credits_raw": credits,
        "keywords": keywords
        })
  else:
    print(f"⚠️ Failed for movie ID: {movie_id}")
  time.sleep(0.2)

print(f"✅ Successfully fetched full data for {len(movie_data)} movies.")

# Step 3: Store in DataFrame
now_playing = pd.DataFrame(movie_data)
```
